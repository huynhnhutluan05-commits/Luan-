```python
def is_prime(n):
    return n > 1 and all(n % i for i in range(2, int(n**0.5) + 1))

def count_vowels(s):
    return sum(ch.lower() in "aeiou" for ch in s)

def average(numbers):
    return sum(numbers) / len(numbers) if numbers else 0

def main():
    print("(a) Kiểm tra số nguyên tố và in danh sách 1→50")
    primes = [n for n in range(1, 51) if is_prime(n)]
    print("Các số nguyên tố:", primes, "\n")

    print("(b) Đếm số nguyên âm trong chuỗi")
    s = "Artificial Intelligence"
    print(f"Chuỗi: '{s}'")
    print("Số nguyên âm:", count_vowels(s), "\n")

    print("(c) Tính trung bình danh sách số")
    nums = [4, 7, 2, 9, 5]
    print("Danh sách:", nums)
    print("Giá trị trung bình:", average(nums))

if __name__ == "__main__":
    main()
```

```python
class Book:
    def __init__(self, book_id, title, author, price):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.price = price
        
    def display_info(self):
        print("Id:", self.book_id)
        print("Tiêu đề:", self.title)
        print("Tác giả:", self.author)
        print("Giá:", self.price)
        print("-" * 30)
        
class Library:
    def __init__(self):
        self.books = []
        
    def add_book(self, book):
        self.books.append(book)
        
    def search_by_author(self, author_name):
        result = []
        for book in self.books:
            if book.author.lower() == author_name.lower():
                result.append(book)
        return result
    
    def get_total_value(self):
        total = 0
        for book in self.books:
            total += book.price
        return total
    
    def display_info(self):
        for book in self.books:
            book.display_info()



book1 = Book(1, "Cho tôi xin một vé đi tuổi thơ", "Nguyễn Nhật Ánh", 50000)
book2 = Book(2, "Kính vạn hoa", "Nguyễn Nhật Ánh", 60000)
book3 = Book(3, "Harry Potter", "J.K. Rowling", 120000)          

lib = Library()  
lib.add_book(book1)
lib.add_book(book2)
lib.add_book(book3)

def display_books_list(books):
    for idx, book in enumerate(books, start=1):
        book.display_info()

# In tất cả sách
print("--Thông tin tất cả sách trong thư viện--")
display_books_list(lib.books)

# In sách của Nguyễn Nhật Ánh
print("--Sách của Nguyễn Nhật Ánh--")
books_by_author = lib.search_by_author("Nguyễn Nhật Ánh")
display_books_list(books_by_author)


print("Tổng giá trị sách trong thư viện:", lib.get_total_value())
```
